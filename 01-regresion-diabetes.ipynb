```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
diabetes=load_diabetes()
X=diabetes.data
y=diabetes.target
X_entrenamiento,X_prueba,y_entrenamiento,y_prueba=train_test_split(X,y,random_state=42,train_size=0.7)
modelo=LinearRegression().fit(X_entrenamiento,y_entrenamiento)
y_pred = modelo.predict(X_prueba)

plt.scatter(y_prueba, y_pred, color='blue', alpha=0.6)
plt.plot([y_prueba.min(), y_prueba.max()], [y_prueba.min(), y_prueba.max()], color='red', linestyle='--')
plt.xlabel("Valores reales")
plt.ylabel("Predicciones")
plt.title("Regresión Lineal - Diabetes")
plt.grid(True)
plt.show()
```
